# Education Outcome Prediction

**Objective:** Build classification models to predict student pass/fail and evaluate feature importance.

**Dataset:** `data/sample_data.csv` (attendance, parent_education, study_hours, score)

---

```
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix

df = pd.read_csv('data/sample_data.csv')
df['pass'] = (df['score'] >= 60).astype(int)
X = df[['attendance','parent_education','study_hours']]
y = df['pass']
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.25, random_state=42)
clf = RandomForestClassifier(random_state=42, n_estimators=100)
clf.fit(X_train, y_train)
preds = clf.predict(X_test)
print(classification_report(y_test, preds))
```

## Feature Importance

```
importances = pd.Series(clf.feature_importances_, index=X.columns).sort_values(ascending=False)
importances
```
